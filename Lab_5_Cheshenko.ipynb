#3. Реалізація алгоритмічного завдання

```python
def bubble_sort_with_count(arr):
    # Контроль правильності введення [cite: 29]
    if not isinstance(arr, list) or not all(isinstance(x, (int, float)) for x in arr):
        raise ValueError("Вхідні дані мають бути списком чисел")

    n = len(arr)
    swaps = 0
    data_copy = arr.copy()

    for i in range(n):
        for j in range(0, n - i - 1):
            if data_copy[j] > data_copy[j + 1]:
                data_copy[j], data_copy[j + 1] = data_copy[j + 1], data_copy[j]
                swaps += 1
    return data_copy, swaps

# Тестові приклади [cite: 28]
test_cases = [
    [64, 34, 25, 12, 22, 11, 90],
    [5, 1, 4, 2, 8],
    [10, 9, 8, 7, 6, 5]
]

for t in test_cases:
    sorted_arr, count = bubble_sort_with_count(t)
    print(f"Оригінал: {t} -> Відсортовано: {sorted_arr}, Перестановок: {count}")
```

#4. Візуалізація результатів

```python
import matplotlib.pyplot as plt

# Дані для графіка
labels = ['Test 1', 'Test 2', 'Test 3']
swap_counts = [bubble_sort_with_count(t)[1] for t in test_cases]

plt.bar(labels, swap_counts, color='red')
plt.title('Кількість перестановок у Bubble Sort')
plt.ylabel('Кількість операцій')
plt.show()
```
